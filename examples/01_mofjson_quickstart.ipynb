# MOFJSON Quickstart Guide

Welcome to **MOFJSON**! This notebook provides a complete run-through of how to use the `mofjson` package to convert, explore, and reconstruct Metal-Organic Frameworks.

MOFJSON represents MOFs in three layers:
1. **Identifiers & Metadata**: Names, properties, and provenance.
2. **Crystal Layer**: Lossless fallback (lattice, fractional coordinates, space group).
3. **Reticular Layer**: LLM-friendly graph representation (topology, metal nodes, linkers).

Let's get started!

```python
import json
from pathlib import Path

# Core MOFJSON imports
from mofjson.convert.cif import MOFFromCif
from mofjson.deconstruct import Deconstruct
from mofjson.convert.pymatgen_bridge import MOFToStructure
```

## 1. Converting a CIF to MOFJSON

We'll start by loading a standard CIF file. We use `MOFFromCif` to parse the crystal structure and derive the periodic bond graph. Then, we use `Deconstruct` to run the graph algorithms that identify the discrete metal nodes, organic linkers, and the overall topology.

```python
# Path to an example CIF file (UiO-66 analog)
cif_path = Path("../tests/data/zr6_uio66like.cif")

# Step 1: Parse CIF into the Crystal layer
print("Parsing CIF...")
doc = MOFFromCif(cif_path, derive_bonds=True)

# Step 2: Deconstruct into the Reticular layer
print("Deconstructing into Reticular layer...")
doc = Deconstruct(doc)

print("\nSuccess! The MOFJSON document is ready.")
```

## 2. Exploring the MOFJSON Document

The resulting `doc` is a Pydantic model (`MOFJSONDocument`). We can easily explore its properties programmatically.

```python
# Explore the Crystal Layer
crystal = doc.crystal
print("--- Crystal Layer ---")
print(f"Space Group: {crystal.symmetry.space_group_symbol}")
print(f"Lattice: a={crystal.lattice.a:.2f}, b={crystal.lattice.b:.2f}, c={crystal.lattice.c:.2f}")
print(f"Number of atoms in unit cell: {crystal.n_atoms}\n")

# Explore the Reticular Layer
reticular = doc.reticular
print("--- Reticular Layer ---")
print(f"Topology (RCSR symbol): {reticular.topology.rcsr_symbol}")
print(f"Number of distinct metal nodes: {len(reticular.nodes)}")
print(f"Number of distinct linkers: {len(reticular.linkers)}\n")

# Inspect the first metal node
node = reticular.nodes[0]
print("--- First Metal Node ---")
print(f"Formula: {node.formula}")
print(f"Archetype: {node.archetype}")
print(f"Connectivity: {node.connectivity}")
print("Metal Sites:")
for site in node.sites:
    print(f"  - {site.element} (Coordination: {site.coordination_number}, Geometry: {site.geometry_symbol})")
```

## 3. Preparing Data for LLMs

When prompting an LLM, providing the full crystal layer (with all fractional coordinates) consumes too many tokens. Instead, we extract just the **reticular layer**, which is highly token-efficient and provides the explicit graph structure the LLM needs for reasoning.

```python
# Dump only the reticular layer to a JSON string
reticular_json = reticular.model_dump_json(indent=2, exclude_none=True)

print("Reticular JSON snippet for LLM prompt:")
print(reticular_json[:500] + "\n... [truncated] ...\n}")
```

## 4. Reconstructing a CIF (Lossless Round-trip)

Because MOFJSON retains the explicit crystal layer, it is completely lossless. We can bridge it back to a `pymatgen` Structure and export it as a CIF file, perfectly reconstructing the original 3D geometry.

```python
# Convert MOFJSON document back to a pymatgen Structure
struct = MOFToStructure(doc)

# Export to CIF format
reconstructed_cif = struct.to(fmt="cif")

print("Reconstructed CIF snippet:")
print("\n".join(reconstructed_cif.splitlines()[:20]))
print("... [truncated] ...")

# You can save it to disk like this:
# with open("reconstructed.cif", "w") as f:
#     f.write(reconstructed_cif)
```
